# **Starter Example Problem**: MIP Optimization
---

## 0. Import PuLP package

```python
from pulp import *
```

## 1. Define the optimization problem: profit maximization

> **syntax**: LpProblem("`name`", `mode (LpMinimize: default / LpMaximize)`)

```python
model = LpProblem("Resource_Allocation_Optimization", LpMaximize)
```

## 2. Define Decision Variables

> **syntax**: LpVariable("`name`", lowBound=`lower bound`, upBound=`upper bound`, cat=`variable category (cont, bin, int)`)

```python
product_A = LpVariable("Product_A", lowBound=0, cat='Integer')
product_B = LpVariable("Prouduct_B", lowBound=0, cat='Integer')
product_C = LpVariable("Prouduct_C", lowBound=0, cat='Integer')
```

## 3. Define the Objective Function

Given that 
- A: \$20 
- B: \$30 
- C: \$40

`Profit = (20 * A) + (30 * B) + (40 * C)`

> **syntax**: model += `objective function`, "`name`"

```python
model += 20*product_A + 30*product_B + 40*product_C, "Total Profit"
```

## 4. Define Constraints

Given that 
- total working hours: **100 hr**
    - labor hour required to produce **A**: 1.0 hr
    - labor hour required to produce **B**: 1.5 hr
    - labor hour required to produce **C**: 1.8 hr
- total machine time: **60 hr**
    - machine time required to produce **A**: 0.5 hr
    - machine time required to produce **B**: 1.0 hr
    - machine time required to produce **C**: 1.2 hr

`Labor Constraint = 1.0 * A + 1.5 * B + 1.8 * C ≤ 100`\
`Machine Constraint = 0.5 * A + 1.0 * B + 1.2 * C ≤ 60`

> **syntax**: model += `constraint` (equality or inequality), "`name`"

```python
model += 1.0*product_A + 1.5*product_B + 1.8*product_C <= 100, "Labor_Constraint"
model += 0.5*product_A + 1.0*product_B + 1.2*product_C <= 60, "Machine_Constraint"
```

## 5. Solve the Model

PuLP comes with CBC solver for general LP problems.

> **syntax**: model.solve()

## 6. Result Presentation

- Determining whether the model reached the optimal solution: **model.status**
    - -3 : Undefined\
      -2 : Unbounded\
      -1 : Infeasible\
       0 : Unsolved\
       1 : Optimal
    - **LpStatus** is an object in a custom class in *PuLP* that acts like a Python dictionary.\
      LpStatus[`index`] prints out the corresponding strings.

> **syntax**: LpStatus[model.status]

- Objective function stored in the model: **model.objective**
    - value() presents the objective function value of the solution.

> **syntax**: value(model.objective)

- The constraints stored in the model: **model.constraints**
    - value(model.constraints[`name of a constraint`]) presents the $LHS - RHS$ value of the indexed constrained when substituting in the solution.
        - e.g. `1.0 * A + 1.5 * B + 1.8 * C - 100`\
          This is because a constraint is stored as `1.0 * A + 1.5 * B + 1.8 * C - 100 ≤ 0`

```python
model.solve()
print(f"Status: {LpStatus[model.status]}")
print(f"Maximum Profit: ${value(model.objective):.2f}")

print("\n--- Optimal Production Quantities ---")

print(f"Product A: {int(value(product_A))} units")
print(f"Product B: {int(value(product_B))} units")
print(f"Product C: {int(value(product_C))} units")

print("\n--- Resource Utilization ---")
print(f"Labor Used: {100 + value(model.constraints['Labor_Constraint']):.2f} hours (Max 100)")
print(f"Macine Time Used: {60 + value(model.constraints['Machine_Constraint']):.2f} hours (Max 60)")
```
